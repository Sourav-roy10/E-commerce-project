```python
grouped_df = (
    df.groupby('region')
    .agg(
        {
            'quantity': 'mean',
            'unit_price': 'mean',
            'discount': 'mean',
            'delivery_days': 'mean',
            'customer_rating': 'mean',
            'revenue': 'sum',
        }
    )
    .sort_values(
        by=['quantity', 'unit_price', 'discount', 'delivery_days', 'customer_rating', 'revenue'],
        ascending=False
    )
)
grouped_df
```

```python
grouped_df2 = (
    df.groupby('payment_method')
    .agg(
        {
            'quantity': 'mean',
            'unit_price': 'mean',
            'discount': 'mean',
            'delivery_days': 'mean',
            'customer_rating': 'mean',
            'revenue': 'sum',
        }
    )
    .sort_values(
        by=['quantity', 'unit_price', 'discount', 'delivery_days', 'customer_rating', 'revenue'],
        ascending=False
    )
)
grouped_df2
```

```python
grouped_df3 = (
    df.groupby('month')
    .agg(
        {
            'quantity': 'mean',
            'unit_price': 'mean',
            'discount': 'mean',
            'delivery_days': 'mean',
            'customer_rating': 'mean',
            'revenue': 'sum',
        }
    )
    .sort_values(
        by=['quantity', 'unit_price', 'discount', 'delivery_days', 'customer_rating', 'revenue'],
        ascending=False
    )
)
grouped_df3
```
